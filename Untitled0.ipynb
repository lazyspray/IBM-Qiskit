````python
# BellGuard — complete single-cell Google Colab project
TEAM_MEMBERS = "KUSHAL, AVYAN, ARUNABH"

import sys
import subprocess
from pathlib import Path
import shutil

# Install dependencies.
subprocess.run(
    [
        sys.executable, "-m", "pip", "install", "-q",
        "qiskit==2.6.0",
        "matplotlib==3.10.8",
        "pylatexenc==2.11",
    ],
    check=True,
)

project = Path("/content/BellGuard")
project.mkdir(parents=True, exist_ok=True)

# Standalone source: also included in the downloaded submission.
source = r'''
import argparse
import csv
import json
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

import qiskit
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
from qiskit.quantum_info import Statevector

SHOTS = 1024
SEED = 42

CASES = [
    ("direct_clean", "Direct measurement: clean", False, False),
    ("direct_phase", "Direct measurement: phase error", True, False),
    ("verified_clean", "Verification: clean", False, True),
    ("verified_phase", "Verification: phase error", True, True),
]


def build_circuit(phase_error=False, verification=False):
    circuit = QuantumCircuit(2)

    # Prepare the entangled Bell pair.
    circuit.h(0)
    circuit.cx(0, 1)
    circuit.barrier()

    # Introduce a controlled phase error.
    if phase_error:
        circuit.z(0)
    circuit.barrier()

    # Reverse preparation to expose the phase difference.
    if verification:
        circuit.cx(0, 1)
        circuit.h(0)

    return circuit


def main():
    parser = argparse.ArgumentParser(description="BellGuard ideal simulation")
    parser.add_argument("--team", default="Enter team members")
    args = parser.parse_args()

    output = Path(__file__).resolve().parent / "results"
    output.mkdir(exist_ok=True)

    circuits = []
    records = []
    bits = ["00", "01", "10", "11"]

    for name, title, error, verification in CASES:
        circuit = build_circuit(error, verification)

        if verification:
            expected = {"01": 1.0} if error else {"00": 1.0}
        else:
            expected = {"00": 0.5, "11": 0.5}

        actual = Statevector.from_instruction(circuit).probabilities_dict()

        for bit in bits:
            if abs(actual.get(bit, 0) - expected.get(bit, 0)) > 1e-10:
                raise AssertionError(f"Probability check failed: {name}, {bit}")

        circuit.measure_all()
        circuits.append(circuit)
        records.append({
            "case": name,
            "title": title,
            "expected_probabilities": expected,
        })

        figure = circuit.draw(output="mpl", fold=-1)
        figure.suptitle(title + " | Ideal simulator", fontsize=12)
        figure.savefig(
            output / f"{name}_circuit.png",
            dpi=160,
            bbox_inches="tight",
        )
        plt.close(figure)

    sampler = StatevectorSampler(seed=SEED)
    sampled = sampler.run(circuits, shots=SHOTS).result()

    figure, axes = plt.subplots(
        2, 2, figsize=(11, 7), layout="constrained"
    )

    for record, result, axis in zip(records, sampled, axes.flat):
        counts = result.data.meas.get_counts()
        record["counts"] = counts

        values = [counts.get(bit, 0) for bit in bits]
        axis.bar(
            bits, values,
            color=["#168AAD", "#F28E2B", "#94A3B8", "#168AAD"],
        )
        axis.set_title(record["title"])
        axis.set_xlabel("Measured bits: q1 q0")
        axis.set_ylabel("Number of shots")
        axis.set_ylim(0, SHOTS * 1.15)

        for position, value in enumerate(values):
            axis.text(
                position, value + 15, str(value),
                ha="center", fontsize=11,
            )

        print(f"{record['title']}: {counts}")

    figure.suptitle(
        "BellGuard — Ideal simulator results\n"
        "Two qubits | 1,024 shots per experiment",
        fontsize=16,
    )
    figure.savefig(output / "comparison.png", dpi=160)
    plt.close(figure)

    report = {
        "project": "BellGuard",
        "team": args.team,
        "backend": "Qiskit StatevectorSampler — ideal classical simulation",
        "qiskit_version": qiskit.__version__,
        "shots_per_experiment": SHOTS,
        "seed": SEED,
        "bit_order": "q1 q0",
        "exact_probability_checks": "All four experiments passed",
        "experiments": records,
    }

    (output / "results.json").write_text(
        json.dumps(report, indent=2), encoding="utf-8"
    )

    with (output / "results.csv").open(
        "w", newline="", encoding="utf-8"
    ) as file:
        writer = csv.writer(file)
        writer.writerow(["Experiment", "00", "01", "10", "11"])
        for record in records:
            writer.writerow(
                [record["title"]]
                + [record["counts"].get(bit, 0) for bit in bits]
            )

    print("\nAll four exact-probability checks PASSED.")
    print("Results are simulated; no quantum hardware was used.")


if __name__ == "__main__":
    main()
'''

(project / "bellguard.py").write_text(source, encoding="utf-8")

requirements = """qiskit==2.6.0
matplotlib==3.10.8
pylatexenc==2.11
"""
(project / "requirements.txt").write_text(requirements, encoding="utf-8")

readme = f"""# BellGuard
## Revealing a hidden quantum phase change

Team members: {TEAM_MEMBERS}

## Project
A two-qubit quantum-computing software demonstration using Python and
Qiskit. It compares direct measurement and inverse verification of a
clean Bell pair and a Bell pair with a deliberately introduced Z error.

## Run
Use Python 3.12, then:
```bash
python -m pip install -r requirements.txt
python bellguard.py
```
"""
(project / "README.md").write_text(readme, encoding="utf-8")

# Run the generated script inside the project directory.
subprocess.run(
    [sys.executable, "bellguard.py", "--team", TEAM_MEMBERS],
    cwd=project,
    check=True,
)

# Create a ZIP archive of the project folder.
shutil.make_archive("/content/BellGuard_Submission", "zip", project)
print("\nBellGuard project successfully executed and archived to 'BellGuard_Submission.zip'.")
````
